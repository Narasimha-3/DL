# PRE-COMPUTED POLICIES 

```python
from PIL import Image
import matplotlib.pyplot as plt
from autoaugment import ImageNetPolicy, CIFAR10Policy, SVHNPolicy, SubPolicy
```

```python
def show_sample(policy):
    image_paths = [r'archive\valid\Car\10158.jpeg', r'archive\valid\Car\10164.jpeg']
    image_set = [Image.open(path) for path in image_paths]
    fig, axes = plt.subplots(nrows=2, ncols=4, figsize=(10, 5))
    imgs = []
    for img in image_set:
        for _ in range(8):
            imgs.append(policy(img))
    for ax, img in zip(axes.flat, imgs):
        ax.imshow(img)
        ax.axis('off') 

    plt.tight_layout()
    plt.show()
```

## ImageNet Policy

```python
show_sample(ImageNetPolicy())
```

## SVHNPolicy

```python
show_sample(SVHNPolicy())
```

## CIFAR10Policy

```python
show_sample(CIFAR10Policy())
```
