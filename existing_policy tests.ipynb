```python
from PIL import Image
import matplotlib.pyplot as plt
from autoaugment import ImageNetPolicy, CIFAR10Policy, SVHNPolicy, SubPolicy
```

```python
def show_sample(policy):
    image_paths = [r'C:\Users\DIVYANSH DESHMUKH\Desktop\Semester\Seventh Semester\Deep Learning\DL\flowers\daisy\5547758_eea9edfd54_n.jpg', r'C:\Users\DIVYANSH DESHMUKH\Desktop\Semester\Seventh Semester\Deep Learning\DL\flowers\daisy\5794835_d15905c7c8_n.jpg', r'flowers/rose/12240303_80d87f77a3_n.jpg']
    image_set = [Image.open(path) for path in image_paths]
    fig, axes = plt.subplots(nrows=2, ncols=4, figsize=(10, 5))
    imgs = []
    for img in image_set:
        for _ in range(8):
            imgs.append(policy(img))
    for ax, img in zip(axes.flat, imgs):
        ax.imshow(img)
        ax.axis('off') 

    plt.tight_layout()
    plt.show()
```

## ImageNet Policy

```python
show_sample(ImageNetPolicy())
```
